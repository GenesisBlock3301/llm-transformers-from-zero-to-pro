```python
# Loading & saving
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained("bert-base-cased")

tokenizer("Using a Transformer network is simple")
```

```python
# Tokenization

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

sequence = "Using a Transformer network is simple and words language languages."
tokens = tokenizer.tokenize(sequence)

print(tokens)
print(tokenizer.vocab.get("language"))
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)

print(ids)
```

```python
# Decoding
decoded_string = tokenizer.decode([7993, 170, 13809, 23763, 2443, 1110, 3014, 1105, 1734, 1846, 3483, 119])
print(decoded_string)
```
