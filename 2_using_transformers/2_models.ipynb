```python
# creating transformer
from transformers import AutoModel

model = AutoModel.from_pretrained("bert-base-cased")
model
```

```python
# Loading & saving model
model.save_pretrained("my_model")
```

```python
# Encoding Text
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

encoded_input = tokenizer("Hello, I'm a single sentence!")
print(encoded_input)
```

```python
# Decode Endcoding
tokenizer.decode(encoded_input["input_ids"])
```

```python
# Batch encoding and decoding
encoded_input = tokenizer(["How are you?", "I'm fine, thank you!"], padding=True, return_tensors="pt")
print(encoded_input)
```

```python
# Truncating Input
encoded_input = tokenizer(
    "This is a very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very long sentence.",
    truncation=True,
)
print(encoded_input["input_ids"])
```

```python
encoded_input = tokenizer(
    ["How are you?", "I'm fine, thank you!"],
    padding=True,
    return_tensors="pt",
    max_length=5,
    truncation=True,
)
print(encoded_input)
```

```python
import torch


sequences = [
    "I've been waiting for a HuggingFace course my whole life.",
    "I hate this so much!",
]

# The manually created `encoded_sequences` had different lengths, leading to the ValueError.
# Transformer models require input sequences within a batch to be of uniform length.
# This is usually handled by padding the shorter sequences.
# Instead of manually encoding and then trying to create a tensor,
# we should use the `tokenizer` (which is already loaded from a previous cell)
# to encode and pad the sequences automatically.

# The `tokenizer` method directly takes the list of strings,
# handles tokenization, padding, and returns a dictionary of PyTorch tensors,
# which is the expected input format for the model.
model_inputs = tokenizer(sequences, padding=True, return_tensors="pt")

# The model expects keyword arguments for its inputs (like input_ids, attention_mask).
# We can unpack the dictionary `model_inputs` using `**`.
output = model(**model_inputs)
print(output)
```
