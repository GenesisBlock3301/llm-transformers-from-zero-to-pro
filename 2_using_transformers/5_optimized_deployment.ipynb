```python
from huggingface_hub import InferenceClient

# এই লাইনটা ঠিক এভাবে লেখো (কোনো মডেল নাম না দিয়ে শুধু URL)
client = InferenceClient("http://127.0.0.1:8080")

# টেস্ট করো
print("টেস্টিং ১...")
print(client.text_generation("বাংলাদেশের রাজধানী কী?", max_new_tokens=20))

print("\nটেস্টিং ২...")
print(client.text_generation("একটা মজার গল্প বলো", max_new_tokens=100))
```
